# Debugging a recycle solve in a notebook

The isomerization loop from `plant.py`: a mixer, a CSTR and a splitter, with 80 % of the effluent recycled.

```python
import runpy, warnings
warnings.simplefilter("ignore")
from difflow import debug

fs = runpy.run_path("plant.py")["fs"]
```

```python
with debug.attach():                 # a difflow-dbg> prompt, under the cell
    fs.solve(acceleration="wegstein")
```
